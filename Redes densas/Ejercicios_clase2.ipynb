## Guardar archivos del proyecto en Google Drive

Esta celda monta Google Drive y copia el proyecto a la carpeta `Ciencia De Datos`. La copia no elimina archivos existentes.

```python
from google.colab import drive, files
from pathlib import Path
import shutil
import zipfile

# Montar Google Drive.
drive.mount('/content/drive')

mi_unidad = Path('/content/drive/MyDrive')
destino = mi_unidad / 'Ciencia De Datos'
destino.mkdir(parents=True, exist_ok=True)

# Si ya existe una copia en Drive, se usa como origen.
origen = mi_unidad / 'Apuntes-Ciencia-de-Datos'
archivos_origen = list(origen.rglob('*')) if origen.exists() else []

if not archivos_origen:
    print('Selecciona un archivo ZIP que contenga todos los archivos del proyecto.')
    archivos_subidos = files.upload()

    if not archivos_subidos:
        raise FileNotFoundError('No se seleccionó ningún archivo ZIP.')

    nombre_zip = next(iter(archivos_subidos))
    ruta_zip = Path('/content') / nombre_zip

    if ruta_zip.suffix.lower() != '.zip':
        raise ValueError('Debes seleccionar un archivo con extensión .zip.')

    with zipfile.ZipFile(ruta_zip, 'r') as archivo_zip:
        archivo_zip.extractall('/content')

    carpetas_proyecto = [
        carpeta for carpeta in Path('/content').iterdir()
        if carpeta.is_dir() and carpeta.name not in {'.config', 'drive', 'sample_data'}
    ]

    if not carpetas_proyecto:
        raise FileNotFoundError('El ZIP no contiene una carpeta de proyecto.')

    origen = carpetas_proyecto[0]

for elemento in origen.iterdir():
    destino_elemento = destino / elemento.name
    if elemento.is_dir():
        shutil.copytree(elemento, destino_elemento, dirs_exist_ok=True)
    else:
        shutil.copy2(elemento, destino_elemento)

archivos_copiados = [elemento for elemento in destino.rglob('*') if elemento.is_file()]
print(f'Archivos guardados en: {destino}')
print(f'Total de archivos en la carpeta destino: {len(archivos_copiados)}')
```
